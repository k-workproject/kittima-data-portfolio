## Note

ธุรกรรมที่ตรวจพบว่าเป็นการฉ้อโกงจะถูกยกเลิก ดังนั้นในการตรวจจับการฉ้อโกง จึงไม่ควรนำคอลัมน์เหล่านี้ (oldbalanceOrg, newbalanceOrig, oldbalanceDest, newbalanceDest) มาใช้งาน

## Description

1. step: หน่วยเวลาในโลกจำลอง
- โดย 1 step เท่ากับ 1 ชั่วโมง รวมทั้งหมด 744 steps (เป็นการจำลองเหตุการณ์ในช่วงเวลา 30 วัน)

2. type: ระเภทของธุรกรรม ได้แก่
  - CASH-IN (การฝากเงิน)
  - CASH-OUT (การถอนเงิน)
  - DEBIT (การหักเงิน)
  - PAYMENT (การชำระเงิน)
  - TRANSFER (การโอนเงิน)

3. amount: จำนวนเงินในธุรกรรมนั้นๆ (หน่วยเป็นสกุลเงินท้องถิ่น)

4. nameOring: ลูกค้าที่เป็นผู้เริ่มต้นทำธุรกรรม

5. oldbalanceOrig: ยอดเงินคงเหลือก่อนทำธุรกรรม

5. newbalanceOrig: ยอดเงินคงเหลือหลังจากทำธุรกรรม

6. nameDest: ลูกค้าที่เป็นผู้รับเงินจากธุรกรรมนั้น

7. oldbalanceDest: ยอดเงินคงเหลือของผู้รับเงินก่อนทำธุรกรรม
(หมายเหตุ: ไม่มีข้อมูลสำหรับลูกค้าที่ชื่อขึ้นต้นด้วยตัวอักษร M ซึ่งหมายถึงร้านค้าหรือ Merchants)

8. newbalanceDest: ยอดเงินคงเหลือของผู้รับเงินหลังจากทำธุรกรรม
(หมายเหตุ: ไม่มีข้อมูลสำหรับลูกค้าที่ชื่อขึ้นต้นด้วยตัวอักษร M ซึ่งหมายถึงร้านค้าหรือ Merchants)

9. isFraud: ระบุว่าธุรกรรมดังกล่าวเป็นการกระทำโดยผู้ไม่หวังดี (fraudulent agents) ภายในระบบจำลองหรือไม่ ในชุดข้อมูลนี้ พฤติกรรมทุจริตคือการที่ผู้ไม่หวังดีเข้าควบคุมบัญชีของลูกค้า แล้วพยายามโอนเงินทั้งหมดออกจากบัญชีไปยังบัญชีอื่น ก่อนจะถอนเงินออกจากระบบไป

10. isFlaggedFraud: ระบบมีการกำหนดเกณฑ์เพื่อควบคุมการโอนเงินจำนวนมากจากบัญชีหนึ่งไปยังอีกบัญชีหนึ่ง โดยจะมีการทำเครื่องหมาย (flag) หากพบความพยายามที่เข้าข่ายผิดกฎหมาย ซึ่งในชุดข้อมูลนี้หมายถึงความพยายามโอนเงินมากกว่า 200,000 หน่วยในการทำธุรกรรมเพียงครั้งเดียว

# Question

## 1. หาบัญชีผู้รับที่รับเงินตั้งแต่ 10 ครั้งขึ้นไปใน 1 ชั่วโมง

```python
%%sql
SELECT
  nameDest,
  step,
  COUNT(*) AS total_txn,
  SUM(amount) AS total_amount
FROM data
WHERE type IN ('TRANSFER', 'CASH_OUT')
GROUP BY nameDest, step
HAVING COUNT(*) >= 10
ORDER BY total_txn DESC
LIMIT 10;
```

```python
%%sql
SELECT
  COUNT(*) AS total_txn,
  SUM(isFraud) AS fraud_txn,
  SUM(isFlaggedFraud) AS flagged_txn
FROM data
WHERE nameDest = 'C1590550415';
```

```python
%%sql
WITH flagged AS (
  SELECT nameDest, step, COUNT(*) AS txn_count
  FROM data
  WHERE type IN ('TRANSFER', 'CASH_OUT')
  GROUP BY nameDest, step
  HAVING COUNT(*) >= 10
)
SELECT
  (SELECT COUNT(*) FROM flagged)         AS flagged_group,
  COUNT(*)                               AS total_txn_in_group,
  ROUND(SUM(d.amount), 2)                AS total_amount,
  SUM(d.isFraud)                         AS fraud_txn_in_group,
  ROUND(100.0 * SUM(d.isFraud) / COUNT(*), 2) AS fraud_rate_pct
FROM data AS d
JOIN flagged AS f
  ON d.nameDest = f.nameDest AND d.step = f.step
WHERE d.type IN ('TRANSFER', 'CASH_OUT');
```

**สรุป**

*จากการตรวจสอบธุรกรรมที่มีการรับเงินตั้งแต่ 10 ครั้งขึ้นไปใน 1 ชั่วโมง พบว่ามีชื่อบัญชีผู้รับที่น่าสงสัยทั้งหมด 76 บัญชี เป็นจำนวน 846 ธุรกรรม รวมเป็นมูลค่าทั้งสิ้น 	381378314.05 ซึ่งเมื่อเทียบกับข้อมูล (isFraud) แล้วไม่พบธุรกรรมที่เป็น Fraud ในข้อมูลชุดนี้*
